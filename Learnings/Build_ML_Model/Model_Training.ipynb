```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Housing.csv')
df
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
ls = ['mainroad','guestroom','basement','hotwaterheating','airconditioning','prefarea','furnishingstatus']
df[ls] = df[ls].replace({'yes':1,'no':0})
df[ls] = df[ls].replace({'unfurnished':0,'semi-furnished':1,'furnished':2})
df[ls] = df[ls].astype(int)
```

```python
df
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
li = ['price','area']
df[li] = scaler.fit_transform(df[li])
df.head()
```

```python
df.dtypes
```

```python
corre = df.corr()
corre

plt.figure(figsize=(12,8))
sns.heatmap(
    corre,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)
plt.show()
```

```python
df.hist(figsize=(10,10),bins=10)
plt.show()
```

```python
x = df.drop('price',axis=1)

y = df['price']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_predict = lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
lr_accuracy = r2_score(y_test,y_predict)*100
```
