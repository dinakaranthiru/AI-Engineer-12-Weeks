##### AI(Artificial Intelligence)

```python
# Making machines perform tasks that normally require human intelligence.
# AI focuses on making machines think, reason, and make decisions like humans.

# Examples:

# Virtual assistants (Copilot, Siri)
# Chess-playing programs
# Recommendation systems
# Self-driving cars
```

##### ML(Machine learning)

```python
# A subset of AI where systems learn from data.
# ML allows computers to learn from data instead of being explicitly programmed.

# Examples:

# Email spam detection
# Stock prediction
# Customer churn prediction
# Fraud detection
```

##### DL(Deep Learning)

```python
# A subset of ML that uses neural networks with many layers
# Deep Learning uses artificial neural networks inspired by the human brain.

# Examples:

# Face recognition
# Voice assistants
# Translation systems
# ChatGPT and Copilot
# Autonomous vehicles
```

##### Types of Machine Learning

* Supervised Learning
* Unsupervised Learning
* Reinforcement Learning
