### Handling Missing values

```python
from sklearn.impute import SimpleImputer

Sp = SimpleImputer(missing_values=np.nan,strategy='mean')
Sp.fit(X[:,1:3])
X[:,1:3] = Sp.transform(X[:,1:3])
```
