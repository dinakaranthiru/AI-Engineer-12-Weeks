## KNN - K-Nearest Neighbors

```python
# K-Nearest Neighbors (KNN) is a simple and popular supervised machine learning algorithm used for both classification and regression problems.
# The basic idea is:

# "Similar data points are likely to belong to the same category."

# When a new data point arrives, KNN finds the K nearest neighbors (existing data points) and makes a prediction based on them.
```

### How KNN Works

```python
# # Step 1: Choose the value of K
# # K represents the number of nearest neighbors to consider.
# # Example:

# # K = 3 → consider the 3 closest points
# # K = 5 → consider the 5 closest points

# # Step 2: Calculate Distance
# # Measure the distance between the new point and all training data points.

# Step 3: Find K Nearest Neighbors
# Sort all distances and select the K smallest distances.

# Step 4: Make Prediction
# For Classification
# Use majority voting.

# For Regression
# Take the average of the K nearest values.
```

### Example

```python
# Example
# Suppose we want to classify whether a fruit is an Apple or Orange.
# Training Data

# Weight (g)         Color Score        Fruit
# 150                    7              Apple
# 170                    8              Apple
# 130                    5              Orange 
# 120                    4              Orange

# New Fruit:
# Weight = 160 g
# Color Score = 7

# If K = 3:
# Nearest neighbors might be:

# If K = 3:
# Nearest neighbors might be:

# Neighbor          Fruit
# 150,7             Apple
# 170,8             Apple
# 130,5             Orange
# Majority = Apple
# Prediction: Apple
```
