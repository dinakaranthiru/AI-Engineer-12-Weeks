### Random Forest Algorithm

```python
# Random Forest is a supervised machine learning algorithm that combines many Decision Trees and makes predictions based on the majority vote (for classification) or average prediction (for regression)
```

### How Random Forest Works

##### Step 1: Create Multiple Decision Trees

```python


# The algorithm randomly selects samples from the dataset and creates many decision trees.
# Example:
# Dataset
#    │
#    ├── Tree 1
#    ├── Tree 2
#    ├── Tree 3
#    ├── Tree 4
#    └── Tree 5
```

##### Step 2: Each Tree Makes a Prediction
 

```python
# For a house:

# Tree     --        Prediction
# Tree1              High Price
# Tree2              High Price
# Tree3              Low Price
# Tree4              High Price 
# Tree5              High Price
```

##### Step 3: Majority Voting

```python

# The final prediction is the class predicted by most trees.
# High Price = 4 votes
# Low Price  = 1 vote

# Final Prediction = High Price
```
