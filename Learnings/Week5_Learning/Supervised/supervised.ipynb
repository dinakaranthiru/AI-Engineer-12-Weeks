##### Supervised Machine Learning algorithm 

```python
#Supervised learning is a type of machine learning algorithm used to predict the output based on labelled data.
#Labelled Data - each input has a correct output
```

```python
# Common Supervised Algorithms
# Classification
# Predicts categories.
# Examples:

# Spam or Not Spam
# Disease or No Disease
# High Price or Low Price House

# Algorithms:

# Random Forest
# Decision Tree
# SVM
# KNN
# Logistic Regression
# Naive Bayes

# Regression
# Predicts numerical values.
# Examples:

# House Price Prediction
# Salary Prediction

# Algorithms:

# Linear Regression
# Random Forest Regressor
# Decision Tree Regressor

# Real-Life Example
# Teacher gives:
# Question + Answer

# Student learns both.
# This is Supervised Learning.
```

### Accuracy:


- Accuracy in classification problem is the number of   correct predictions made by the model divided by the total number of predictions.

### Precision:



- Precision is defined as the number of true positive divided by the   number of true positive + number of false positive.

### Recall:


- Recall express the ability to find all relevant instances in a dataset 

### F1-Score

- in case where we want to find an optimal blend of precision and recall we can combine the two metrics using what is called F1 Score.
