##### Supervised Machine Learning algorithm 

```python
#Supervised learning is a type of machine learning algorithm used to predict the output based on labelled data.
#Labelled Data - each input has a correct output
```

```python
# Common Supervised Algorithms
# Classification
# Predicts categories.
# Examples:

# Spam or Not Spam
# Disease or No Disease
# High Price or Low Price House

# Algorithms:

# Random Forest
# Decision Tree
# SVM
# KNN
# Logistic Regression
# Naive Bayes

# Regression
# Predicts numerical values.
# Examples:

# House Price Prediction
# Salary Prediction

# Algorithms:

# Linear Regression
# Random Forest Regressor
# Decision Tree Regressor

# Real-Life Example
# Teacher gives:
# Question + Answer

# Student learns both.
# This is Supervised Learning.
```
