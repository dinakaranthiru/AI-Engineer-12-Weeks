### Unsupervised Learning

```python
# In Unsupervised Learning, only input data is given.
# There are no labels or correct answers.
# The algorithm finds hidden patterns by itself.
```

```python
# Main Types of Unsupervised Learning
# 1. Clustering
# Groups similar data points together.

# Purpose:
# Find natural clusters/segments in data.
# Popular algorithms:

#### K-Means:
# You choose number of clusters k.
# Algorithm assigns each point to one of k clusters.
# Often used for customer segmentation, document clustering.

#### Hierarchical Clustering:
# Builds a tree (dendrogram) of clusters.
# You can cut the tree at different levels to get different numbers of clusters.

#### DBSCAN:
# Finds clusters based on density of points.
# Can automatically detect noise/outliers.
# Useful when clusters are not spherical.

#### K Nearset Neighbour Clustering

#### PCA 
 
#### LDA

```
