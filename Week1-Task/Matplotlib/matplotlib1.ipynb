# Bar charts = compare categories of data by representing each category with a bar

```python
cat = ["Grains",'Fruits','vegetables','Protiens','Dairy','Sweets']
values = np.array([4,3,2,5,3,1])

# plt.bar(cat,values,color = 'red')
#For horizontal barchart we have to use the barh
plt.barh(cat,values,color = 'red')

plt.title("Daily Consumption")
plt.xlabel('Food')
plt.show()
```

# Pie chart = circular chart divided into slices to show percentage of the total.

```python
categories = ["Freshman","Sophomores","juniors","Seniors"]
values = np.array([300,250,275,225])
colors = ['red','blue','green','yellow']

plt.pie(values,labels = categories,autopct = "%1.1f%%",colors = colors,explode=[0,0.05,0,0])
plt.show()
```

# Scatter plot =  shows the relationship between two variables , helps to identify a correlation (+,-,None)
Example: Study hours vs Test Scores

```python
x1 = [0,1,1,2,3,4,5,6,7,7,8]
y1 = [55,60,65,62,68,70,75,78,82,85,87]

x2= [0,1,2,2,3,4,5,6,7,8,8]
y2 = [50,58,65,70,72,78,84,88,92,95,97]

plt.scatter(x1,y1,color = 'red',alpha=0.5,s=100,label = "Class A")
plt.scatter(x2,y2,color = 'blue',alpha=0.5,s=100,label = "Class B")
plt.xlabel("Hours studied")
plt.ylabel("Grade")
plt.legend()
plt.title("Test Scores")
```
