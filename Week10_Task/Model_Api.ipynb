### Week 10 — Build AI API

#### Task
Convert the trained machine learning model into an **API service** using:

- Flask

#### Endpoint Example

**POST /predict**

#### Input
```json
{
  "text": "cannot login to system"
}
```

#### Output
```json
{
  "category": "Login Issue"
}
```

#### Deliverables

- Create an API endpoint (`POST /predict`) for predictions.
- Accept input data in JSON format.
- Return the predicted category as a JSON response.
- Test the API with sample requests.

#### Skills Tested

- Model Deployment
- API Development
